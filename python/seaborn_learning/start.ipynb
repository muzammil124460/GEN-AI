# Lineplot

```python
import seaborn as sb
import matplotlib.pyplot as mtpl
```

```python
data = sb.load_dataset("tips")
data.head()
```

```python

sb.lineplot(x="total_bill",y="tip", data=data , hue="sex" ,palette="deep",sort=True ,style="sex" ,markers=True)
mtpl.show()
```

# regplot

```python
sb.regplot(data=data,x="total_bill",y="tip",ci=95,scatter_kws={
    "s":80,
    "alpha":0.6,
    "color":"green",
    "marker":"^"
},  line_kws={
        "linewidth":3,
        "color":"red",
        "linestyle":"--"
    },order=2)
```

# Histogram

```python
sb.histplot(data=data,x="total_bill",y="tip",)
```

# Kernel Density Estimation Plot

```python
sb.kdeplot(data=data,x="total_bill",fill=True)
```

# Matrix Plots

```python

tips = sb.load_dataset("tips")
data = tips.corr(numeric_only=True)
sb.heatmap(data=data,annot=True,cmap="viridis",linewidths=10,cbar=True,vmin=0)
```
