```python
from torchvision import datasets , transforms
import torch
```

```python
transform_data = transforms.ToTensor()

training_data = datasets.MNIST(
    root="./data",
    train=True,
    transform=transform_data,
    download=True
)

testing_data = datasets.MNIST(
    root="./data",
    train=True,
    transform=transform_data,
    download=True
)
```

```python
image, label = training_data[0]

print(image.shape)
print(label)
print(type(label))
```

### DATASET AND DATALOADER

```python
from torch.utils.data import Dataset,DataLoader

# Class of Data set

class ImageClass(Dataset):

    def __init__(self,features,label):
        self.feature = torch.tensor(data=features,dtype=torch.float32).reshape(-1,1,28,28),
        self.label = torch.tensor(label,dtype=torch.log)
    
    def __len__(self):
        return len(self.feature)

    def __getitem__(self, index):
        return self.feature[index],self.label[index]
```

### DataLOADERS

```python
data_for_training = DataLoader(training_data,batch_size=32,shuffle=True)
print(len(data_for_training))
data_for_testing = DataLoader(testing_data,batch_size=32,shuffle=False)
```

LAYERS AND FILTERS

```python
from torch import nn
```

```python
class MODEL(nn.Module):

    def __init__(self,features_size):
        super().__init__()

        self.filters = nn.Sequential(
            nn.Conv2d(features_size,16,3,padding="same"),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2,stride=2),

             nn.Conv2d(16,32,3,padding="same"),
              nn.ReLU(),
              nn.MaxPool2d(kernel_size=2,stride=2)
            
        )

        self.network = nn.Sequential(
            nn.Flatten(),
            nn.Linear(32*7*7,64),
            nn.ReLU(),
            nn.Linear(64,10),
            
       
        )
    def forward(self,x):
        x = self.filters(x)
        x = self.network(x)
        return x
```

### Training

```python

learning_rate = 0.01
epochs = 1
model = MODEL(1)
loss_fun = nn.CrossEntropyLoss()

opitmize = torch.optim.Adam(
    model.parameters(),
    lr=learning_rate
)
```

```python
for epoch in range(epochs):

    for batch, (features, label) in enumerate(data_for_training):

        output = model(features)

        loss_data = loss_fun(output, label)

        opitmize.zero_grad()
        loss_data.backward()
        opitmize.step()

        if (batch + 1) % 100 == 0:
            print(
                f"Epoch {epoch+1} | "
                f"Batch {batch+1}/{len(data_for_training)} | "
                f"Loss {loss_data.item():.4f}"
            )
```

```python
correct = 0
total = 0

model.eval()

with torch.no_grad():

    for features, label in data_for_testing:

        output = model(features)

        prediction = output.argmax(dim=1)

        total += label.size(0)
        correct += (prediction == label).sum().item()

accuracy = 100 * correct / total

print(f"Test Accuracy: {accuracy:.2f}%")
```
