```python
import pandas as pd


email_data = pd.read_csv("email.csv")

email_data
```

```python

email_data.value_counts

email_data= email_data[email_data["Category"] != '{"mode":"full"' ]
email_data["Category"] = email_data["Category"].map(lambda x :1 if x == "spam" else 0)
email_data.value_counts
```

```python

X = transform.fit_transform(email_data["Message"])
Y = email_data["Category"]

X_train,x_test,Y_train,y_test = train_test_split(X,Y,test_size=0.3)

from sklearn.linear_model import LogisticRegression

linear = LogisticRegression()

model = linear.fit(X_train,Y_train)

predicted_output = model.predict(x_test)


data = [
    "spam" if x == 1 else "ham"
    for x in predicted_output[0:10]
]

print(data)
print(y_test[0:10].map(lambda x :"spam" if x == 1 else "ham"))

```

```python
import gensim
print(gensim.__version__)
```

```python
sentences = [
    ["i", "love", "machine", "learning"],
    ["i", "love", "deep", "learning"],
    ["machine", "learning", "is", "useful"],
    ["deep", "learning", "is", "powerful"]
]
```

```python
from gensim.models import Word2Vec

model = Word2Vec(
    sentences,
    vector_size=100,
    window=5,
    workers=3,
    min_count=1
)
```

```python
similarity = model.wv.similarity("machine", "learning")

print(similarity)
```
