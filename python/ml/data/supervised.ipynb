```python
import pandas as pd
```

```python
data = pd.read_csv("../email.csv")


final_data = data[data["Category"].isin(["spam","ham"])]

x = final_data["Message"]

y = final_data["Category"].map({
    "ham":0
    ,"spam":1
})

```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vector_transfer = TfidfVectorizer()

x_ = vector_transfer.fit_transform(x)

print(x_.shape)
```

```python
from sklearn.model_selection import train_test_split

x_train ,X_test ,y_train,Y_test = train_test_split(
    x_,y
    ,test_size=0.2
)

y_train
```

```python
from sklearn.linear_model import LogisticRegression



model = LogisticRegression()

final_model = model.fit(x_train,y_train)

output =final_model.predict(X_test)
print(output[0:6])
Y_test.head()
```

```python
email = ["Congratulations! You have won a $1,000 cash prize. Click here now to claim your reward."]

temail = vector_transfer.transform(email)

output = final_model.predict(temail)
output

```
