```python
import pandas as pd
from sklearn.preprocessing import StandardScaler , MinMaxScaler






```

# StandardScaler

```python
data = {
    "salary":[1000,2000,3000,4000,5000]
}
data = pd.DataFrame(data)

stnd = StandardScaler()

data = stnd.fit_transform(data)
data
```

# MinMeanScaler

```python
data = {
    "salary":[1000,2000,3000,4000,5000]
}
data = pd.DataFrame(data)

minmeanscaler = MinMaxScaler()
data = minmeanscaler.fit_transform(data)
data
```
