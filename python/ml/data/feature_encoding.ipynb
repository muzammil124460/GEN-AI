```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder,LabelEncoder,OrdinalEncoder
```

# One Hot encoding

```python
encoder = OneHotEncoder()
```

```python
color = [
    "red",
    "yellow",
    "green",
    "red",
    "green",
    "yellow"
]
```

```python
data =pd.DataFrame(color,columns=["color"])
encoded = encoder.fit_transform(data)
modified_data = encoded.toarray()
k = pd.DataFrame(modified_data,columns=encoder.get_feature_names_out())
k
```

```python
final_data = pd.concat([data,k],axis=1)
final_data
```

# LableEncoder

```python
encoder = LabelEncoder()
pd_data =encoder.fit_transform(data["color"])
print(pd_data)
```

```python
demodata = {
    "size": ["Small", "Large", "Medium", "Small"]
}

data = pd.DataFrame(demodata)

encoder = OrdinalEncoder(categories=[["Small", "Medium", "Large"]])
data_mm = encoder.fit_transform(data[["size"]])
data_mm

modified = pd.DataFrame(data_mm,columns=["order"])
final_data =pd.concat([data,modified],axis=1)
final_data
```
