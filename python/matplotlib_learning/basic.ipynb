```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [ 10,20,30,40,50]

plt.plot(x,y ,color = "red" , marker = "o" , linestyle="--")
plt.xlabel("x label")
plt.ylabel("y label")
plt.title("practice")
plt.legend("&")

plt.show()
```

# Scatter Plot

```python
x = [1,2,3,4,5]
y = [10,20,30,40,50]
plt.scatter(x,y)
plt.show()
```

# Histogram

```python
marks = [45, 50, 55, 60, 62, 65, 70, 72, 75, 80, 85, 90]
```

```python
plt.hist(marks , bins=5,
    color="skyblue",
    edgecolor="black")
```

# Pie Chart

```python
students = [40, 30, 20, 10]
languages = ["Python", "JavaScript", "Java", "C++"]
explode = [0.1, 0.1, 0.1, 0.1]

plt.pie(students,labels=languages,autopct="%1.1f%%",explode=explode)
plt.show()
```
